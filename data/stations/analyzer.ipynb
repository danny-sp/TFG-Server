## Columns Analysis

https://openchargemap.org/develop/api#/schemas/POI

Columns found in the data:
- DataProvider: Source of the data. Not relevant for our system.
- **OperatorInfo: Information about the operator of the charging station. Relevant**
- UsageType: Type of usage (e.g., public, private). We assumme all are selectable. Not relevant for our system.
- **StatusType: Operational status of the station. We discard those that are not operational (e.g., "Disabled", "Inactive")**
- SubmissionStatus: Status of the data submission. Not relevant for our system.
- UserComments: User comments about the station. Not relevant for our system.
- PercentageSimilarity: Null value. Not relevant for our system.
- MediaItems: Media items related to the station. Not relevant for our system.
- IsRecentlyVerified: Whether the station has been recently verified. Not relevant for our system.
- DateLastVerified: Date of the last verification. Not relevant for our system.
- **ID: Unique identifier for the station. Relevant.**
- UUID: Unique identifier for the station. Not relevant for our system.
- ParentChargePointID: Null value. Not relevant for our system.
- DataProviderID: Source of the data. Not relevant for our system.
- DataProvidersReference: Identifier for the operator. Not relevant for our system.
- OperatorID: Identifier for the operator. Not relevant for our system.
- OperatorsReference: Identifier for the operator. Not relevant for our system.
- UsageTypeID: Reference for the type of usage. Not relevant for our system.
- **UsageCost: Cost of using the station. Relevant (if available)**
- **AddressInfo: Information about the station's location. Relevant**
- **Connections: Information about the station's different connection types. Relevant**
- **NumberOfPoints: Number of charging points available at the station. Relevant**
- GeneralComments: General comments about the station. Not relevant for our system.
- DatePlanned: Null value. Not relevant for our system.
- DateLastConfirmed: Null value. Not relevant for our system.
- StatusTypeID: Reference for the operational status. Not relevant for our system.
- DateLastStatusUpdated: Date of the last status update. Not relevant for our system.
- MetadataValues: Null value (>99% of rows). Not relevant for our system.
- DataQualityLevel: Unused value. Not relevant for our system.
- DateCreated: Date of creation of the station entry. Not relevant for our system.
- SubmissionStatusTypeID: Reference for the submission status. Not relevant for our system.

# Data Cleaning

```python
operators_dict = {
    2247: "Iberdrola",
    207: "Endesa",
    91: "Repsol",
    3499: "Mercadona",
    3377: "Etecnic",
}
```

```python
clean_data = clean_data[
    clean_data["StatusType"].apply(lambda x: isinstance(x, dict) and x.get("Title") == "Operational")
]

print("Not operational:", len(df_data) - len(clean_data))
```

```python
print(clean_data["AddressInfo"].iloc[0].keys())
```

```python
data_to_insert = []

for station in raw_data:
    try:
        id = station["ID"]
        name = station["AddressInfo"]["Title"]
        lat = station["AddressInfo"]["Latitude"]
        lon = station["AddressInfo"]["Longitude"]
        conn = station["Connections"]
    except Exception as e:
        continue

    if station["OperatorID"] in operators_dict:
        operator = operators_dict[station["OperatorID"]]
    else:
        operator = "Unknown"

    data_to_insert.append({
        "name": name,
        "lat": lat,
        "lon": lon,
        "operator": operator,
        "connections": conn,
    })

with open('stations_data_to_insert.json', 'w', encoding='utf-8') as f:
    json.dump(data_to_insert, f, indent=4, ensure_ascii=False)

len(data_to_insert)
```
